## Análise Exploratória

### Consolidação e tipagem


```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = pd.read_csv('Housing.csv')
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df.info()
```

```python
#Convertendo a Para Metros Quadrados

converter = ['sqft_living' , 'sqft_lot', 'sqft_above', 'sqft_basement', 'sqft_living15', 'sqft_lot15']

for i in converter:
    df[i]= df[i] * 0.092903
```

```python
df['zipcode'] = df['zipcode'].astype(str)
```

```python
df.info()
```

```python
df.head()
```

```python
df.isna().sum()
```

```python

traducao_colunas = {
    'id': 'id_imovel',
    'date': 'data',
    'price': 'preco',
    'bedrooms': 'quartos',
    'bathrooms': 'banheiros',
    'sqft_living': 'm2_area_habitavel',    
    'sqft_lot': 'm2_terreno',
    'floors': 'andares',
    'waterfront': 'vista_mar_rio',         
    'view': 'nota_vista',                  
    'condition': 'condicao_imovel',        
    'grade': 'nota_design',                
    'sqft_above': 'm2_acima_solo',          
    'sqft_basement': 'm2_porao',          
    'yr_built': 'ano_construcao',
    'yr_renovated': 'ano_renovacao',
    'zipcode': 'codigo_postal',
    'lat': 'latitude',
    'long': 'longitude',
    'sqft_living15': 'm2_area_vizinhos_15', 
    'sqft_lot15': 'm2_terreno_vizinhos_15' 
}

df_traduzido = df.rename(columns=traducao_colunas)

print(df_traduzido.columns)
```

## Fase 2

### Análise histórica de preços

```python
df_traduzido['ano'] = df_traduzido['data'].dt.year
df_traduzido['mes'] = df_traduzido['data'].dt.month
```

```python
df_traduzido.head()
```

```python
df_media_anual = df_traduzido.groupby('ano')['preco'].mean().reset_index().round(2)
```

```python
df_media_anual
```

```python
media_mensal = df_traduzido.groupby(['ano','mes'])['preco'].mean().reset_index().round(2)
media_mensal
```

```python
df_2015 = df_traduzido[df_traduzido['ano'] == 2015]

df_media_mensal_2015  = df_2015.groupby('mes')['preco'].mean().reset_index().round(2)
```

```python
df_media_mensal_2015
```

```python
fig = px.line(df_media_mensal_2015 , x = 'mes' , y='preco', markers = True, title = 'Média de Preços Mensais 2015')
fig.show()
```

### Análise Localizada

```python
df_traduzido['codigo_postal'].value_counts().reset_index().head()
```

```python
bairro_escolhido = 98103

df_bairro = df_traduzido[df_traduzido['codigo_postal'] == bairro_escolhido]

print(f'O preço médio do cep {bairro_escolhido:.2f} é R${df_bairro['preco'].mean():.2f}, o maior preco é {df_bairro['preco'].max():.2f} o menor é {df_bairro['preco'].min()} ')
```

## Ranking de Bairros

```python
media_preco_bairro = (df_traduzido.groupby('codigo_postal')['preco'].mean()
.sort_values(ascending=False)
.reset_index()
.rename(columns= {'preco' : 'Media_preco'})
.round(2)
)
media_preco_bairro
```

```python
top5 = media_preco_bairro.head(5)
top5
```

```python
fig_top5 = px.bar(
    top5,
    x = 'Media_preco',
    y = 'codigo_postal',
    orientation='h',
    text_auto = '.3s',
    color = 'Media_preco',
    title ='Top 5 bairros com maiores preços médios'
)
fig.update_yaxes(type = 'category')
fig_top5.show()
```
